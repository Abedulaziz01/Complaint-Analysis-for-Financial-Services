```python

import pandas as pd

# Read the CSV file in chunks of 1000 rows
df_iter = pd.read_csv('../data/filtered.csv', chunksize=1000)

# Combine all chunks into one DataFrame
df = pd.concat(df_iter, ignore_index=True)

# Now df contains the entire dataset
print("Shape of the data:", df.shape)
```

```python


# Drop rows where 'Consumer complaint narrative' is missing or empty
df = df[df['Consumer complaint narrative'].notna() & (df['Consumer complaint narrative'].str.strip() != '')]

# Confirm how many rows remain
print(f"Remaining complaints with narratives: {df.shape[0]}")

# Overwrite the same CSV file
df.to_csv("../data/filtered.csv", index=False)
print("Updated CSV saved (only complaints with narratives).")
```

```python
# lowering to lower case 

# Lowercase specific columns
df['Product'] = df['Product'].str.lower()
df['Consumer complaint narrative'] = df['Consumer complaint narrative'].str.lower()

# Check the result
print(df[['Product', 'Consumer complaint narrative']].head())

# Save back to the same CSV file
df.to_csv("../data/filtered.csv", index=False)
print("Text has been lowercased and saved back to the file.")
```

```python

import re


# Function to clean text by removing special characters and boilerplate
def clean_text(text):
    if isinstance(text, str):
        # Remove special characters (punctuation, symbols)
        text = re.sub(r'[^\w\s]', '', text)
        
        # Remove common boilerplate phrases (case insensitive)
        text = re.sub(r'(Dear\s+[A-Za-z\s,]+|Dear\s*(Sir/Madam|Customer|[A-Za-z]+).*?)(\n|$)', '', text, flags=re.IGNORECASE)
        text = re.sub(r'I am writing to complain.*?(\n|$)', '', text, flags=re.IGNORECASE)
        text = re.sub(r'Thank you for your attention.*?(\n|$)', '', text, flags=re.IGNORECASE)
        text = re.sub(r'Please note that.*?(\n|$)', '', text, flags=re.IGNORECASE)
        text = re.sub(r'Regards,\s*.*?(\n|$)', '', text, flags=re.IGNORECASE)
        text = re.sub(r'Best regards,\s*.*?(\n|$)', '', text, flags=re.IGNORECASE)
        
        # Remove extra spaces
        text = re.sub(r'\s+', ' ', text).strip()  # Replace multiple spaces with a single one
        
    return text

# Clean the 'Consumer complaint narrative' column
df['Consumer complaint narrative'] = df['Consumer complaint narrative'].apply(clean_text)

# Optional: Check the result for the first few rows
print(df[['Product', 'Consumer complaint narrative']].head())

# Save the cleaned data back to the same CSV
df.to_csv('../data/filtered.csv', index=False)
print("Special characters and boilerplate removed. Data saved back to the file.")
```

```python

# Count missing values per column
missing = df.isnull().sum().sort_values(ascending=False)

# Optional: Show as percentage too
missing_percent = (missing / len(df) * 100).round(2)

# Combine into a single table
missing_report = pd.DataFrame({
    'Missing Values': missing,
    'Percent Missing': missing_percent
})

print("Missing values summary:\n")
print(missing_report)
```

```python
# missing value analysis 

# Drop sparse columns with low value for analysis
df.drop(columns=[
    'Consumer disputed?',      # 94% missing — not useful
    'Tags',                    # 80% missing — vague categories
    'Company public response'  # 52% missing — not needed for fairness insights
], inplace=True)

# Fill missing 'State' with "Unknown"
df['State'] = df['State'].fillna('Unknown')  # 0.49% missing

# (Optional) Fill missing 'Sub-product' and 'Sub-issue' with "Unknown" if you want to keep all rows
df['Sub-product'] = df['Sub-product'].fillna('Unknown')  # 5.28% missing
df['Sub-issue'] = df['Sub-issue'].fillna('Unknown')      # 12.99% missing

# Save the updated file back to CSV
df.to_csv('../data/filtered.csv', index=False)
print("Columns dropped and missing values filled. File updated.")
```
